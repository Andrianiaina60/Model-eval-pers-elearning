# Cellule 5 — Extraire RandomForest et Dummy sur N=125 / N=105

```python
# ============================================================
# CELLULE 5 — Validation des performances RandomForest
# ============================================================

def resume_modele(
    df,
    n_cohorte,
    modele
):

    temp = df[
        (df["FRAMEWORK"] == "M_T1")
        &
        (df["MODELE"] == modele)
    ].copy()


    assert len(temp) > 0, (
        f"{modele} absent"
    )


    return {
        "COHORTE":
            n_cohorte,

        "MODELE":
            modele,

        "N_EVALUATIONS":
            len(temp),

        "MAE_MOYEN":
            temp["MAE"].mean(),

        "MAE_STD":
            temp["MAE"].std(),

        "RMSE_MOYEN":
            temp["RMSE"].mean(),

        "RMSE_STD":
            temp["RMSE"].std(),

        "R2_MOYEN":
            temp["R2"].mean(),

        "R2_STD":
            temp["R2"].std()
    }


validation_A = pd.DataFrame([

    resume_modele(
        RESULTATS_N125,
        125,
        "RandomForest"
    ),

    resume_modele(
        RESULTATS_N125,
        125,
        "Dummy"
    ),

    resume_modele(
        RESULTATS_N105,
        105,
        "RandomForest"
    ),

    resume_modele(
        RESULTATS_N105,
        105,
        "Dummy"
    )

])


display(
    validation_A.round(4)
)
```

# Cellule 6 — Mesurer le gain par rapport au Dummy

```python
# ============================================================
# CELLULE 6 — Gain RandomForest vs Dummy
# ============================================================

rf = validation_A[
    validation_A["MODELE"] == "RandomForest"
].set_index("COHORTE")

dummy = validation_A[
    validation_A["MODELE"] == "Dummy"
].set_index("COHORTE")


gain_vs_dummy = pd.DataFrame({

    "COHORTE": [125, 105],

    "GAIN_MAE": [
        dummy.loc[125, "MAE_MOYEN"]
        - rf.loc[125, "MAE_MOYEN"],

        dummy.loc[105, "MAE_MOYEN"]
        - rf.loc[105, "MAE_MOYEN"]
    ],

    "GAIN_RMSE": [
        dummy.loc[125, "RMSE_MOYEN"]
        - rf.loc[125, "RMSE_MOYEN"],

        dummy.loc[105, "RMSE_MOYEN"]
        - rf.loc[105, "RMSE_MOYEN"]
    ],

    "GAIN_R2": [
        rf.loc[125, "R2_MOYEN"]
        - dummy.loc[125, "R2_MOYEN"],

        rf.loc[105, "R2_MOYEN"]
        - dummy.loc[105, "R2_MOYEN"]
    ]
})


display(
    gain_vs_dummy.round(4)
)
```

# Cellule 8 — Validation automatique de la Phase A

```python
# ============================================================
# CELLULE 8 — Validation automatique de la Phase A
# ============================================================

rf_125 = rf.loc[125]
rf_105 = rf.loc[105]

dummy_125 = dummy.loc[125]
dummy_105 = dummy.loc[105]


# ------------------------------------------------------------
# 1. Construire les contrôles
# ------------------------------------------------------------

validation_checks = pd.DataFrame(
    [
        {
            "COHORTE": 125,
            "METRIQUE": "MAE",
            "RF": rf_125["MAE_MOYEN"],
            "DUMMY": dummy_125["MAE_MOYEN"],
            "GAIN_RF": (
                dummy_125["MAE_MOYEN"]
                -
                rf_125["MAE_MOYEN"]
            )
        },

        {
            "COHORTE": 125,
            "METRIQUE": "RMSE",
            "RF": rf_125["RMSE_MOYEN"],
            "DUMMY": dummy_125["RMSE_MOYEN"],
            "GAIN_RF": (
                dummy_125["RMSE_MOYEN"]
                -
                rf_125["RMSE_MOYEN"]
            )
        },

        {
            "COHORTE": 125,
            "METRIQUE": "R2",
            "RF": rf_125["R2_MOYEN"],
            "DUMMY": dummy_125["R2_MOYEN"],
            "GAIN_RF": (
                rf_125["R2_MOYEN"]
                -
                dummy_125["R2_MOYEN"]
            )
        },

        {
            "COHORTE": 105,
            "METRIQUE": "MAE",
            "RF": rf_105["MAE_MOYEN"],
            "DUMMY": dummy_105["MAE_MOYEN"],
            "GAIN_RF": (
                dummy_105["MAE_MOYEN"]
                -
                rf_105["MAE_MOYEN"]
            )
        },

        {
            "COHORTE": 105,
            "METRIQUE": "RMSE",
            "RF": rf_105["RMSE_MOYEN"],
            "DUMMY": dummy_105["RMSE_MOYEN"],
            "GAIN_RF": (
                dummy_105["RMSE_MOYEN"]
                -
                rf_105["RMSE_MOYEN"]
            )
        },

        {
            "COHORTE": 105,
            "METRIQUE": "R2",
            "RF": rf_105["R2_MOYEN"],
            "DUMMY": dummy_105["R2_MOYEN"],
            "GAIN_RF": (
                rf_105["R2_MOYEN"]
                -
                dummy_105["R2_MOYEN"]
            )
        }
    ]
)


# ------------------------------------------------------------
# 2. Vérification de la validité numérique
# ------------------------------------------------------------

colonnes_numeriques = [
    "RF",
    "DUMMY",
    "GAIN_RF"
]


assert (
    np.isfinite(
        validation_checks[
            colonnes_numeriques
        ].to_numpy()
    )
    .all()
), (
    "Valeur non finie détectée "
    "dans les métriques de validation."
)


# ------------------------------------------------------------
# 3. Un gain positif signifie :
#
# MAE  : RF < Dummy
# RMSE : RF < Dummy
# R2   : RF > Dummy
# ------------------------------------------------------------

validation_checks[
    "VALIDE"
] = (
    validation_checks[
        "GAIN_RF"
    ] > 0
)


validation_checks[
    "CONTROLE"
] = (
    "RF_meilleur_Dummy_"
    +
    validation_checks[
        "METRIQUE"
    ]
    +
    "_N"
    +
    validation_checks[
        "COHORTE"
    ].astype(str)
)


validation_checks = (
    validation_checks[
        [
            "CONTROLE",
            "COHORTE",
            "METRIQUE",
            "RF",
            "DUMMY",
            "GAIN_RF",
            "VALIDE"
        ]
    ]
)


# ------------------------------------------------------------
# 4. Affichage complet
# ------------------------------------------------------------

print(
    "=== VALIDATION RANDOMFOREST vs DUMMY ==="
)


display(
    validation_checks.round(4)
)


# ------------------------------------------------------------
# 5. Identifier explicitement les contrôles échoués
# ------------------------------------------------------------

controles_echoues = (
    validation_checks[
        ~validation_checks[
            "VALIDE"
        ]
    ]
    .copy()
)


if len(
    controles_echoues
) > 0:

    print(
        "\nPhase A NON validée."
    )

    print(
        "Contrôles échoués :"
    )

    display(
        controles_echoues.round(4)
    )

    raise RuntimeError(
        "RandomForest M_T1 ne surpasse pas "
        "le Dummy sur tous les contrôles. "
        "Vérifier les sorties du Notebook 03 "
        "avant de poursuivre l'analyse XAI."
    )


# ------------------------------------------------------------
# 6. Conclusion
# ------------------------------------------------------------

print(
    "\nPhase A validée : "
    "RandomForest M_T1 surpasse le Dummy "
    "sur MAE, RMSE et R², "
    "pour N=125 et N=105."
)
```

# Cellule 16 — Préprocesseur final

```python
# ============================================================
# CELLULE 16 — Préprocesseur final
# ============================================================

def creer_onehot():

    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            drop="first",
            sparse_output=False
        )

    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            drop="first",
            sparse=False
        )


def creer_preprocesseur(
    variables_num,
    variables_cat
):

    pipeline_num = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])


    pipeline_cat = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            creer_onehot()
        )
    ])


    return ColumnTransformer([
        (
            "num",
            pipeline_num,
            variables_num
        ),
        (
            "cat",
            pipeline_cat,
            variables_cat
        )
    ])


print("✅ Préprocesseur défini")
```

# Cellule 17 — Pipeline RandomForest final
On reprend la famille d'hyperparamètres utilisée dans l'étape 03.

```python
# ============================================================
# CELLULE 17 — Pipeline final RandomForest
# ============================================================

preprocessor_final = creer_preprocesseur(
    FEATURES_MT1_NUM,
    FEATURES_MT1_CAT
)


pipeline_rf_final = Pipeline([

    (
        "preprocess",
        preprocessor_final
    ),

    (
        "model",
        RandomForestRegressor(
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    )

])


PARAM_GRID_RF_FINAL = {

    "model__n_estimators":
        [200, 400],

    "model__max_depth":
        [3, 5, None],

    "model__min_samples_leaf":
        [2, 5],

    "model__max_features":
        ["sqrt", 1.0]

}


print("✅ Pipeline RandomForest final construit")
```

#  Cellule 18 — Recherche des hyperparamètres finaux

# Cellule 23 — Regrouper les One-Hot par variable originale

```python
# ============================================================
# CELLULE 23 — Importance par variable originale
# ============================================================

def retrouver_variable_originale(
    feature_transformee
):

    # Numériques
    if feature_transformee.startswith("num__"):

        return feature_transformee.replace(
            "num__",
            "",
            1
        )

    # Catégorielles One-Hot
    if feature_transformee.startswith("cat__"):

        nom = feature_transformee.replace(
            "cat__",
            "",
            1
        )

        # On cherche d'abord les noms les plus longs
        # pour éviter les mauvaises correspondances
        for col in sorted(
            FEATURES_MT1_CAT,
            key=len,
            reverse=True
        ):

            if (
                nom == col
                or
                nom.startswith(col + "_")
            ):

                return col

    return feature_transformee


importance_rf[
    "VARIABLE_ORIGINALE"
] = (
    importance_rf[
        "FEATURE_TRANSFORMEE"
    ]
    .apply(
        retrouver_variable_originale
    )
)


importance_variables = (
    importance_rf

    .groupby(
        "VARIABLE_ORIGINALE",
        as_index=False
    )

    .agg(
        IMPORTANCE=("IMPORTANCE", "sum")
    )

    .sort_values(
        "IMPORTANCE",
        ascending=False
    )

    .reset_index(drop=True)
)


importance_variables[
    "IMPORTANCE_PCT"
] = (
    importance_variables["IMPORTANCE"]
    * 100
)


display(
    importance_variables.head(20)
)
```

# Cellule 24 — Permutation importance en validation croisée


```python
# ============================================================
# CELLULE 24 — Permutation importance avec tuning interne
# ============================================================

from sklearn.base import clone


cv_xai = KFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)


permutation_folds = []


for fold, (
    train_idx,
    test_idx
) in enumerate(
    cv_xai.split(
        X_125
    ),
    start=1
):

    X_train = X_125.iloc[
        train_idx
    ].copy()

    X_test = X_125.iloc[
        test_idx
    ].copy()

    y_train = y_125.iloc[
        train_idx
    ].copy()

    y_test = y_125.iloc[
        test_idx
    ].copy()


    inner_cv = KFold(
        n_splits=4,
        shuffle=True,
        random_state=
            RANDOM_STATE + fold
    )


    recherche_fold = GridSearchCV(

        estimator=clone(
            pipeline_rf_final
        ),

        param_grid=
            PARAM_GRID_RF_FINAL,

        scoring=
            "neg_mean_absolute_error",

        cv=inner_cv,

        n_jobs=-1,

        refit=True
    )


    recherche_fold.fit(
        X_train,
        y_train
    )


    modele_fold = (
        recherche_fold
        .best_estimator_
    )


    perm = permutation_importance(

        modele_fold,

        X_test,
        y_test,

        scoring=
            "neg_mean_absolute_error",

        n_repeats=30,

        random_state=
            RANDOM_STATE + fold,

        n_jobs=-1
    )


    for (
        feature,
        moyenne,
        std
    ) in zip(
        FEATURES_MT1,
        perm.importances_mean,
        perm.importances_std
    ):

        permutation_folds.append({
            "FOLD":
                fold,

            "VARIABLE":
                feature,

            "IMPORTANCE_MAE":
                moyenne,

            "IMPORTANCE_STD":
                std
        })


PERMUTATION_CV = pd.DataFrame(
    permutation_folds
)
```

# Cellule 32 — Explication locale d’un apprenant

```python
# ============================================================
# CELLULE 32 — Fonction d'explication locale SHAP
# ============================================================

def expliquer_apprenant_shap(
    row_position,
    top_n=10
):

    valeurs = shap_values.values[
        row_position
    ]

    base_value = float(
        np.asarray(
            shap_values.base_values[
                row_position
            ]
        ).reshape(-1)[0]
    )

    prediction = float(
        MODELE_RF_FINAL.predict(
            X_125.iloc[
                [row_position]
            ]
        )[0]
    )

    score_reel = float(
        y_125.iloc[
            row_position
        ]
    )


    explication = pd.DataFrame({

        "VARIABLE":
            X_SHAP.columns,

        "VALEUR_SHAP":
            valeurs,

        "ABS_SHAP":
            np.abs(
                valeurs
            )

    })


    explication = (
        explication
        .sort_values(
            "ABS_SHAP",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )


    print(
        "Position :",
        row_position
    )

    print(
        "Valeur de base :",
        round(
            base_value,
            3
        )
    )

    print(
        "Prédiction T2 :",
        round(
            prediction,
            3
        )
    )

    print(
        "T2 réel :",
        round(
            score_reel,
            3
        )
    )

    print(
        "Erreur :",
        round(
            prediction
            -
            score_reel,
            3
        )
    )


    display(
        explication
    )


    return explication
```

## Analyse additionnelle de H3 — Concordance des niveaux aux trois évaluations

Cette comparaison sur les **mêmes 125 apprenants** est exploratoire : T0 comporte cinq questions dans chaque domaine, T1/T2 quinze questions propres au parcours. Les seuils 40/70 sont utilisés **uniquement pour l'analyse de concordance**, et ne remplacent pas les niveaux relatifs K-means du moteur de recommandation.

```python
from sklearn.metrics import cohen_kappa_score

niveaux_ord = ["Debutant", "Intermediaire", "Avance"]
def etiqueter_niveau(v):
    x = float(v)
    if not np.isfinite(x) or not 0 <= x <= 100:
        return pd.NA
    return "Debutant" if x < 40 else ("Intermediaire" if x < 70 else "Avance")

base_h3 = BASE_125.copy()
assert len(base_h3) == 125
assert base_h3["_ID_NORM"].is_unique
assert base_h3["T1__PARCOURS"].isin(["DA","BI","DS","IA"]).all()
# Correspondance score T0 / parcours T1 connu ; aucun parcours T2 utilisé
base_h3["SCORE_T0_DANS_PARCOURS_SUIVI"] = np.select(
    [base_h3["T1__PARCOURS"].eq(p) for p in ["DA","BI","DS","IA"]],
    [base_h3[f"T0__S_{p}"] for p in ["DA","BI","DS","IA"]],
    default=np.nan
)
for etape, col in [("T0", "SCORE_T0_DANS_PARCOURS_SUIVI"), ("T1","T1__SCORE"), ("T2","TARGET_T2")]:
    base_h3[f"NIVEAU_{etape}_40_70"] = base_h3[col].map(etiqueter_niveau)
assert base_h3[[f"NIVEAU_{e}_40_70" for e in ["T0","T1","T2"]]].notna().all().all()
res_h3 = []
for depart in ("T0", "T1"):
    a = base_h3[f"NIVEAU_{depart}_40_70"]
    b = base_h3["NIVEAU_T2_40_70"]
    res_h3.append({"ETAPE_SOURCE": depart, "N": len(base_h3),
                   "ACCORD_BRUT": float((a == b).mean()),
                   "KAPPA_PONDERE_LINEAIRE": float(cohen_kappa_score(a, b, labels=niveaux_ord, weights="linear"))})
resultats_h3 = pd.DataFrame(res_h3)
resultats_h3.to_csv(TABLE_DIR / "H3_concordance_niveaux_40_70.csv", index=False, encoding="utf-8-sig")
base_h3[["_ID_NORM","T1__PARCOURS", "NIVEAU_T0_40_70", "NIVEAU_T1_40_70", "NIVEAU_T2_40_70"]].to_csv(
    TABLE_DIR / "H3_niveaux_apparies_N125.csv", index=False, encoding="utf-8-sig")
display(resultats_h3)
print("H3 : association descriptive ; vérifier la comparabilité des instruments et l'incertitude avant conclusion.")
```
