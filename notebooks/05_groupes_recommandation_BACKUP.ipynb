# 05 — Groupes et recommandation pédagogique

## Objectifs

Cette étape construit les mécanismes de recommandation pédagogique à
partir des résultats des étapes précédentes.

Deux moments sont distingués :

\[
R_{T0}: Profil + T0 \rightarrow Parcours
\]

pour recommander un parcours parmi DA, BI, DS et IA ;

et

\[
R_{T1}: Profil + T0 + T1 + \widehat{T2}
\rightarrow Niveau + Lacunes + Prérequis + Groupe
\]

pour constituer des groupes pédagogiques homogènes.

Les recommandations sont produites pour des groupes d'apprenants
présentant des besoins compatibles. Elles ne constituent pas une
décision pédagogique individuelle automatique.

La décision finale reste soumise à validation humaine.

# Cellule 9 — Encoder les niveaux déclarés

```python
# ============================================================
# CELLULE 9 — Encodage des niveaux déclarés
# ============================================================

NIVEAU_ORDINAL = {
    "Aucun": 0,
    "Debutant": 1,
    "Intermediaire": 2,
    "Avance": 3
}


BASE_RECO_T0["LVL_INFORMATIQUE"] = (
    BASE_RECO_T0["Niveau informatique"]
    .map(NIVEAU_ORDINAL)
)

BASE_RECO_T0["LVL_EXCEL"] = (
    BASE_RECO_T0["Niveau Excel"]
    .map(NIVEAU_ORDINAL)
)

BASE_RECO_T0["LVL_POWER_BI"] = (
    BASE_RECO_T0["Niveau Power BI"]
    .map(NIVEAU_ORDINAL)
)

BASE_RECO_T0["LVL_PYTHON"] = (
    BASE_RECO_T0["Niveau Python"]
    .map(NIVEAU_ORDINAL)
)

BASE_RECO_T0["LVL_IA"] = (
    BASE_RECO_T0["Niveau IA"]
    .map(NIVEAU_ORDINAL)
)


print("✅ Niveaux déclarés encodés")
```

# Cellule 11 — Préparation des prérequis déclarés
Ce mapping est utilisé uniquement comme départage secondaire, pas comme vérité absolue.

```python
# ============================================================
# CELLULE 11 — Readiness / prérequis déclarés
# ============================================================

BASE_RECO_T0["READY_DA"] = (
    BASE_RECO_T0[
        [
            "LVL_INFORMATIQUE",
            "LVL_EXCEL"
        ]
    ]
    .mean(axis=1)
)


BASE_RECO_T0["READY_BI"] = (
    BASE_RECO_T0[
        [
            "LVL_INFORMATIQUE",
            "LVL_EXCEL",
            "LVL_POWER_BI"
        ]
    ]
    .mean(axis=1)
)


BASE_RECO_T0["READY_DS"] = (
    BASE_RECO_T0[
        [
            "LVL_INFORMATIQUE",
            "LVL_PYTHON"
        ]
    ]
    .mean(axis=1)
)


BASE_RECO_T0["READY_IA"] = (
    BASE_RECO_T0[
        [
            "LVL_INFORMATIQUE",
            "LVL_PYTHON",
            "LVL_IA"
        ]
    ]
    .mean(axis=1)
)


print("✅ Indicateurs de prérequis calculés")
```

Cellule 12 — Fonction de recommandation

```python
# ============================================================
# CELLULE 12 — Recommandation de parcours
# ============================================================

PARCOURS = [
    "DA",
    "BI",
    "DS",
    "IA"
]


SCORE_COL = {
    "DA": "S_DA",
    "BI": "S_BI",
    "DS": "S_DS",
    "IA": "S_IA"
}


READY_COL = {
    "DA": "READY_DA",
    "BI": "READY_BI",
    "DS": "READY_DS",
    "IA": "READY_IA"
}


def recommander_parcours(row):

    # --------------------------------------------------------
    # 1. Scores T0
    # --------------------------------------------------------

    scores = {
        p: float(row[SCORE_COL[p]])
        for p in PARCOURS
    }


    score_max = max(
        scores.values()
    )


    candidats = [
        p
        for p in PARCOURS
        if scores[p] == score_max
    ]


    # --------------------------------------------------------
    # Cas 1 : meilleur T0 unique
    # --------------------------------------------------------

    if len(candidats) == 1:

        choisi = candidats[0]

        return pd.Series({
            "PARCOURS_RECOMMANDE":
                choisi,

            "PARCOURS_CANDIDATS":
                choisi,

            "REGLE_DECISION":
                "T0_unique",

            "SCORE_T0_RECOMMANDE":
                score_max,

            "NB_CANDIDATS":
                1
        })


    # --------------------------------------------------------
    # 2. Départage par objectif professionnel
    # --------------------------------------------------------

    parcours_objectif = row[
        "PARCOURS_OBJECTIF"
    ]


    if (
        pd.notna(parcours_objectif)
        and
        parcours_objectif in candidats
    ):

        return pd.Series({
            "PARCOURS_RECOMMANDE":
                parcours_objectif,

            "PARCOURS_CANDIDATS":
                " | ".join(candidats),

            "REGLE_DECISION":
                "T0_egalite_objectif",

            "SCORE_T0_RECOMMANDE":
                score_max,

            "NB_CANDIDATS":
                len(candidats)
        })


    # --------------------------------------------------------
    # 3. Départage par prérequis déclarés
    # --------------------------------------------------------

    readiness = {
        p: row[READY_COL[p]]
        for p in candidats
    }


    readiness_max = max(
        readiness.values()
    )


    candidats_ready = [
        p
        for p in candidats
        if readiness[p] == readiness_max
    ]


    if len(candidats_ready) == 1:

        choisi = candidats_ready[0]

        return pd.Series({
            "PARCOURS_RECOMMANDE":
                choisi,

            "PARCOURS_CANDIDATS":
                " | ".join(candidats),

            "REGLE_DECISION":
                "T0_egalite_prerequis",

            "SCORE_T0_RECOMMANDE":
                score_max,

            "NB_CANDIDATS":
                len(candidats)
        })


    # --------------------------------------------------------
    # 4. Ne pas forcer une décision
    # --------------------------------------------------------

    return pd.Series({
        "PARCOURS_RECOMMANDE":
            "A_VALIDER",

        "PARCOURS_CANDIDATS":
            " | ".join(candidats_ready),

        "REGLE_DECISION":
            "egalite_persistante",

        "SCORE_T0_RECOMMANDE":
            score_max,

        "NB_CANDIDATS":
            len(candidats_ready)
    })
```

# Cellule 15 — Mesurer l’ambiguïté T0
Comme T0 n’a que 5 questions par domaine, les égalités peuvent être fréquentes.

```python
# ============================================================
# CELLULE 15 — Ambiguïté de la recommandation
# ============================================================

def marge_t0(row):

    scores = sorted(
        [
            row["S_DA"],
            row["S_BI"],
            row["S_DS"],
            row["S_IA"]
        ],
        reverse=True
    )

    return (
        scores[0]
        -
        scores[1]
    )


BASE_RECO_T0[
    "MARGE_T0"
] = (
    BASE_RECO_T0
    .apply(
        marge_t0,
        axis=1
    )
)


resume_marge = (
    BASE_RECO_T0[
        "MARGE_T0"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "MARGE_T0"
    )
    .reset_index(
        name="N"
    )
)


display(
    resume_marge
)
```

#  Cellule 16 — Comparaison descriptive avec le parcours observé à T1

Très important : T1__PARCOURS n’est pas considéré comme la vérité absolue. On mesure seulement la concordance avec le parcours réellement observé.

```python
# ============================================================
# CELLULE 16 — Concordance descriptive avec T1
# ============================================================

ID = "_ID_NORM"


validation_reco = (
    BASE_RECO_T0[
        [
            ID,
            "PARCOURS_RECOMMANDE",
            "PARCOURS_CANDIDATS",
            "REGLE_DECISION",
            "MARGE_T0"
        ]
    ]
    .merge(
        BASE_MT1_125[
            [
                ID,
                "T1__PARCOURS"
            ]
        ],
        on=ID,
        how="inner",
        validate="one_to_one"
    )
)


validation_reco[
    "CONCORDANCE_T1"
] = (
    validation_reco[
        "PARCOURS_RECOMMANDE"
    ]
    ==
    validation_reco[
        "T1__PARCOURS"
    ]
)


# Exclure A_VALIDER pour calculer la concordance
validation_decidee = (
    validation_reco[
        validation_reco[
            "PARCOURS_RECOMMANDE"
        ] != "A_VALIDER"
    ]
)


print(
    "N comparaison :",
    len(validation_reco)
)

print(
    "N recommandations décidées :",
    len(validation_decidee)
)

print(
    "Concordance descriptive avec T1 :",
    round(
        validation_decidee[
            "CONCORDANCE_T1"
        ].mean() * 100,
        2
    ),
    "%"
)
```

# Cellule 17 — Le parcours T1 appartient-il au groupe des meilleurs candidats T0 ?

```python
# ============================================================
# CELLULE 17 — T1 appartient-il aux candidats T0 ?
# ============================================================

def t1_dans_candidats(row):

    candidats = str(
        row["PARCOURS_CANDIDATS"]
    ).split(" | ")

    return (
        row["T1__PARCOURS"]
        in candidats
    )


validation_reco[
    "T1_DANS_CANDIDATS"
] = (
    validation_reco
    .apply(
        t1_dans_candidats,
        axis=1
    )
)


print(
    "T1 présent parmi les candidats T0 :",
    round(
        validation_reco[
            "T1_DANS_CANDIDATS"
        ].mean() * 100,
        2
    ),
    "%"
)
```

# Cellule 18 — Concordance selon la règle utilisée

```python
# ============================================================
# CELLULE 18 — Concordance selon la règle
# ============================================================

concordance_par_regle = (
    validation_reco[
        validation_reco[
            "PARCOURS_RECOMMANDE"
        ] != "A_VALIDER"
    ]
    .groupby(
        "REGLE_DECISION",
        as_index=False
    )
    .agg(
        N=("CONCORDANCE_T1", "size"),
        CONCORDANCE_PCT=(
            "CONCORDANCE_T1",
            lambda x: x.mean() * 100
        )
    )
)


display(
    concordance_par_regle.round(2)
)
```

# Cellule 19 — Concordance selon la marge T0

```python
# ============================================================
# CELLULE 19 — Concordance selon la marge T0
# ============================================================

concordance_par_marge = (
    validation_reco[
        validation_reco[
            "PARCOURS_RECOMMANDE"
        ] != "A_VALIDER"
    ]
    .groupby(
        "MARGE_T0",
        as_index=False
    )
    .agg(
        N=("CONCORDANCE_T1", "size"),
        CONCORDANCE_PCT=(
            "CONCORDANCE_T1",
            lambda x: x.mean() * 100
        )
    )
)


display(
    concordance_par_marge.round(2)
)
```

# Cellule 22 — Recommandation multicritère

```python
# ============================================================
# CELLULE 22 — Recommandation multicritère transparente
# ============================================================

PARCOURS = ["DA", "BI", "DS", "IA"]

SCORE_COL = {
    "DA": "S_DA",
    "BI": "S_BI",
    "DS": "S_DS",
    "IA": "S_IA"
}

READY_COL = {
    "DA": "READY_DA",
    "BI": "READY_BI",
    "DS": "READY_DS",
    "IA": "READY_IA"
}


def recommandation_multicritere(row):

    preuves = {
        p: 0
        for p in PARCOURS
    }

    raisons = {
        p: []
        for p in PARCOURS
    }

    # --------------------------------------------------------
    # 1. Critère T0
    # --------------------------------------------------------

    scores_t0 = {
        p: row[SCORE_COL[p]]
        for p in PARCOURS
    }

    max_t0 = max(scores_t0.values())

    meilleurs_t0 = [
        p
        for p in PARCOURS
        if scores_t0[p] == max_t0
    ]

    for p in meilleurs_t0:
        preuves[p] += 1
        raisons[p].append("meilleur_T0")


    # --------------------------------------------------------
    # 2. Critère objectif professionnel
    # --------------------------------------------------------

    objectif = row["PARCOURS_OBJECTIF"]

    if pd.notna(objectif) and objectif in PARCOURS:
        preuves[objectif] += 1
        raisons[objectif].append("objectif_professionnel")


    # --------------------------------------------------------
    # 3. Critère prérequis
    # --------------------------------------------------------

    readiness = {
        p: row[READY_COL[p]]
        for p in PARCOURS
    }

    max_ready = max(readiness.values())

    meilleurs_ready = [
        p
        for p in PARCOURS
        if readiness[p] == max_ready
    ]

    for p in meilleurs_ready:
        preuves[p] += 1
        raisons[p].append("prerequis")


    # --------------------------------------------------------
    # 4. Parcours avec le plus de preuves
    # --------------------------------------------------------

    max_preuves = max(preuves.values())

    candidats = [
        p
        for p in PARCOURS
        if preuves[p] == max_preuves
    ]


    if len(candidats) == 1:
        principal = candidats[0]
    else:
        principal = "A_VALIDER"


    classement = sorted(
        PARCOURS,
        key=lambda p: (
            preuves[p],
            scores_t0[p],
            readiness[p]
        ),
        reverse=True
    )


    return pd.Series({

        "PARCOURS_PRINCIPAL":
            principal,

        "PARCOURS_CANDIDATS":
            " | ".join(candidats),

        "CLASSEMENT_PARCOURS":
            " > ".join(classement),

        "NB_PREUVES_MAX":
            max_preuves,

        "NB_CANDIDATS_MAX":
            len(candidats),

        "PREUVES_DA":
            preuves["DA"],

        "PREUVES_BI":
            preuves["BI"],

        "PREUVES_DS":
            preuves["DS"],

        "PREUVES_IA":
            preuves["IA"],

        "RAISONS_DA":
            " | ".join(raisons["DA"]),

        "RAISONS_BI":
            " | ".join(raisons["BI"]),

        "RAISONS_DS":
            " | ".join(raisons["DS"]),

        "RAISONS_IA":
            " | ".join(raisons["IA"])
    })
```

## CYCLE 1 — T0 : recommandation, validation humaine et groupes de besoins **proposés**

Ce bloc préserve les 305 diagnostics, y compris les cas `A_VALIDER`. Il sépare explicitement :

- la **proposition algorithmique de parcours** (individuelle),
- la **validation facultative documentée de l'enseignant** (aucune valeur simulée),
- le **choix réel de l'apprenant** (jamais inféré à partir de T1 ou T2),
- les **12 groupes potentiels** (4 parcours × 3 niveaux), calculés ici à titre de *préparation pédagogique avant T1*.

Les ressources au cycle T0 sont proposées **aux groupes**, pas personnalisées individuellement. Les groupes T0 sans décision définitive sont des groupes provisoires et ne constituent pas la population du test expérimental M_T1.

**Politique cible de validation asynchrone :** délai configurable de 3 jours à compter de la notification effectivement documentée. À l’expiration, le processus peut proposer les parcours à l’apprenant sans validation implicite ; les avis tardifs sont archivés séparément. Aucune date de notification n’est reconstruite à partir de T1/T2. Dans les données historiques, cette fonctionnalité reste une simulation de logique métier, pas une validation prospective observée.


```python
from collections import Counter

ETAPE_RECO = "T0"
CYCLES_DIR = TABLE_DIR / "deux_cycles"
CYCLES_DIR.mkdir(parents=True, exist_ok=True)
PARCOURS_CYCLES = ("DA", "BI", "DS", "IA")

def classer_score_t0(score):
    if pd.isna(score):
        return pd.NA
    x = float(score)
    if not (0 <= x <= 100):
        raise ValueError(f"Score T0 invalide : {score}")
    return "Debutant" if x < 40 else ("Intermediaire" if x < 70 else "Avance")

# Une ligne par apprenant ; un cas A_VALIDER reste sans parcours présumé.
reco_t0_cycles = BASE_RECO_MULTI.copy()
assert len(reco_t0_cycles) == 305 and reco_t0_cycles[ID].is_unique
reco_t0_cycles["PARCOURS_AUTO"] = reco_t0_cycles["PARCOURS_PRINCIPAL"].where(
    reco_t0_cycles["PARCOURS_PRINCIPAL"].isin(PARCOURS_CYCLES), pd.NA)
reco_t0_cycles["STATUT_AUTO"] = np.where(
    reco_t0_cycles["PARCOURS_AUTO"].notna(), "PROPOSITION_UNIVOQUE", "A_VALIDER")
reco_t0_cycles["ETAPE_LACUNE"] = "T0"

# Fichier de saisie manuelle : jamais écraser une décision authentique.
f_decision = CYCLES_DIR / "decisions_parcours_enseignant_apprenant_T0.csv"
colonnes_decision = [ID, "DECISION_ENSEIGNANT", "PARCOURS_ENSEIGNANT", "PARCOURS_CHOISI",
                    "COMMENTAIRE_ENSEIGNANT", "VALIDATEUR", "DATE_VALIDATION",
                    "DATE_NOTIFICATION_ENSEIGNANT", "DATE_CHOIX_APPRENANT", "SOURCE_CHOIX_APPRENANT"]
if not f_decision.exists():
    gabarit = reco_t0_cycles[[ID]].copy()
    for c in colonnes_decision[1:]:
        gabarit[c] = ""
    gabarit.to_csv(f_decision, index=False, encoding="utf-8-sig")
    print("Fichier enseignant/apprenant créé sans décision présumée :", f_decision)
decisions = pd.read_csv(f_decision, dtype=str).fillna("")
# Compatibilité ascendante avec les CSV manuels existants : jamais les réécrire.
for c in colonnes_decision:
    if c not in decisions.columns and c != ID:
        decisions[c] = ""
assert set(colonnes_decision).issubset(decisions.columns)
assert set(decisions["DECISION_ENSEIGNANT"].str.upper().str.strip()) <= {
    "", "EN_ATTENTE", "VALIDE", "MODIFIE", "REJETE", "ARBITRE", "EXPIRE_SANS_AVIS", "AVIS_TARDIF"
}, "Statut de décision enseignant non reconnu"
assert decisions[ID].is_unique
ids_inconnus = set(decisions[ID]) - set(reco_t0_cycles[ID])
assert not ids_inconnus, f"IDs étrangers dans le fichier de décision : {len(ids_inconnus)}"
valides_parcours = set(PARCOURS_CYCLES) | {""}
for nom in ("PARCOURS_ENSEIGNANT", "PARCOURS_CHOISI"):
    assert set(decisions[nom].str.strip()) <= valides_parcours, f"Parcours invalide : {nom}"
reco_t0_cycles = reco_t0_cycles.merge(
    decisions[colonnes_decision], on=ID, how="left", validate="one_to_one")
for c in colonnes_decision[1:]:
    reco_t0_cycles[c] = reco_t0_cycles[c].fillna("").astype(str).str.strip()
# Politique de revue asynchrone : pas de date de notification => pas d'expiration déduite.
# Les observations historiques ne sont pas traitées comme une notification réelle.
DELAI_AVIS_ENSEIGNANT_JOURS = 3  # Paramètre de conception, non mesuré expérimentalement.
DATE_REFERENCE_REVUE = pd.Timestamp.now(tz="UTC")  # Horloge à l'exécution seulement.
notifie = pd.to_datetime(reco_t0_cycles["DATE_NOTIFICATION_ENSEIGNANT"], errors="coerce", utc=True)
date_avis = pd.to_datetime(reco_t0_cycles["DATE_VALIDATION"], errors="coerce", utc=True)
echeance = notifie + pd.to_timedelta(DELAI_AVIS_ENSEIGNANT_JOURS, unit="D")
reco_t0_cycles["DATE_ECHEANCE_ENSEIGNANT"] = echeance.dt.strftime("%Y-%m-%dT%H:%M:%SZ").fillna("")
decision_formateur = reco_t0_cycles["DECISION_ENSEIGNANT"].str.upper()
enseignant_renseigne = decision_formateur.isin(["VALIDE", "MODIFIE", "REJETE", "ARBITRE"])
trace_avis = (reco_t0_cycles["VALIDATEUR"].ne("") & date_avis.notna())
avis_exploitable = enseignant_renseigne & trace_avis
avis_tardif = avis_exploitable & echeance.notna() & date_avis.gt(echeance)
expire_sans_avis = (
    echeance.notna()
    & (echeance <= DATE_REFERENCE_REVUE)
    & ~avis_exploitable
)
reco_t0_cycles["STATUT_REVUE_ENSEIGNANT"] = np.select(
    [avis_tardif, avis_exploitable, expire_sans_avis, notifie.notna()],
    ["AVIS_TARDIF", "AVIS_DANS_DELAI_OU_ECHEANCE_INCONNUE", "EXPIRE_SANS_AVIS", "EN_ATTENTE"],
    default="NOTIFICATION_NON_DOCUMENTEE")
# Une proposition IA n'est jamais assimilée à un avis de formateur.
choix_apprenant_saisi = reco_t0_cycles["PARCOURS_CHOISI"].isin(PARCOURS_CYCLES)
parcours_formateur_saisi = reco_t0_cycles["PARCOURS_ENSEIGNANT"].isin(PARCOURS_CYCLES)
enseignant_modifie = avis_exploitable & ~avis_tardif & decision_formateur.isin(["MODIFIE", "ARBITRE"])
enseignant_rejette = avis_exploitable & ~avis_tardif & decision_formateur.eq("REJETE")
parcours_pour_preparation = reco_t0_cycles["PARCOURS_AUTO"].copy()
parcours_pour_preparation.loc[enseignant_modifie | enseignant_rejette] = pd.NA
parcours_pour_preparation.loc[enseignant_modifie & parcours_formateur_saisi] = (
    reco_t0_cycles.loc[enseignant_modifie & parcours_formateur_saisi, "PARCOURS_ENSEIGNANT"])
parcours_pour_preparation.loc[choix_apprenant_saisi] = (
    reco_t0_cycles.loc[choix_apprenant_saisi, "PARCOURS_CHOISI"])
reco_t0_cycles["PARCOURS_PREPARE"] = parcours_pour_preparation
reco_t0_cycles["STATUT_PREPARATION"] = np.select(
    [choix_apprenant_saisi,
     enseignant_rejette,
     enseignant_modifie & parcours_formateur_saisi & reco_t0_cycles["PARCOURS_AUTO"].isna(),
     enseignant_modifie & parcours_formateur_saisi,
     enseignant_modifie,
     reco_t0_cycles["PARCOURS_AUTO"].notna()],
    ["CHOIX_APPRENANT_DOCUMENTE", "REJET_ENSEIGNANT_SANS_CHOIX",
     "ARBITRAGE_FORMATEUR_CAS_AMBIGU", "PROPOSITION_MODIFIEE_PAR_FORMATEUR",
     "MODIFICATION_SANS_PARCOURS", "PROPOSITION_IA_PROVISOIRE_SANS_AVIS"],
    default="PARCOURS_NON_RESOLU")
# Le libre choix reste obligatoire pour finaliser le parcours, même après expiration.
assert reco_t0_cycles[ID].is_unique and len(reco_t0_cycles) == 305

# Lacunes du diagnostic T0 : déjà calculées à partir des QCM du notebook 02.
colonnes_lacunes = [f"LACUNES_{p}" for p in PARCOURS_CYCLES]
assert set([ID] + colonnes_lacunes).issubset(DIAG_T0.columns)
reco_t0_cycles = reco_t0_cycles.merge(
    DIAG_T0[[ID] + colonnes_lacunes], on=ID, how="left", validate="one_to_one")
reco_t0_cycles["SCORE_T0_PARCOURS"] = np.select(
    [reco_t0_cycles["PARCOURS_PREPARE"].eq(p) for p in PARCOURS_CYCLES],
    [reco_t0_cycles[f"S_{p}"] for p in PARCOURS_CYCLES], default=np.nan)
reco_t0_cycles["NIVEAU_DIAGNOSTIQUE_T0"] = reco_t0_cycles["SCORE_T0_PARCOURS"].map(classer_score_t0)
reco_t0_cycles["ID_GROUPE_T0"] = np.where(
    reco_t0_cycles["PARCOURS_PREPARE"].isin(PARCOURS_CYCLES),
    "T0_" + reco_t0_cycles["PARCOURS_PREPARE"].fillna("") + "_" +
    reco_t0_cycles["NIVEAU_DIAGNOSTIQUE_T0"].fillna(""), "")

# Les cas indécis restent dans la table des recommandations mais PAS dans un groupe forcé.
valides_t0_cycles = reco_t0_cycles.loc[
    reco_t0_cycles["PARCOURS_PREPARE"].isin(PARCOURS_CYCLES)].copy()
assert len(valides_t0_cycles) + int(reco_t0_cycles["PARCOURS_PREPARE"].isna().sum()) == 305

def lacunes_les_plus_frequentes(groupe, n=3):
    parcours = groupe["PARCOURS_PREPARE"].iloc[0]
    observations = []
    for cellule in groupe[f"LACUNES_{parcours}"].dropna().astype(str):
        observations += [s.strip() for s in cellule.split(" | ") if s.strip() and not s.startswith("NON-REPONSE")]
    return " | ".join([s for s, _ in Counter(observations).most_common(n)]) if observations else ""

lignes_groupes_t0 = []
for (p,n), bloc in valides_t0_cycles.groupby(["PARCOURS_PREPARE", "NIVEAU_DIAGNOSTIQUE_T0"]):
    lac = lacunes_les_plus_frequentes(bloc)
    lignes_groupes_t0.append({
        "ID_GROUPE": f"T0_{p}_{n}", "PARCOURS": p, "NIVEAU": n,
        "STATUT_NIVEAU": "DESCRIPTIF_T0_40_70", "N_APPRENANTS": bloc[ID].nunique(),
        "LACUNE_PRINCIPALE": lac, "SOURCE_LACUNES": "T0", "ETAPE_LACUNE": "T0",
        "REQUETE_RAG": lac if lac else pd.NA,
        "STATUT_PROPOSITION": "PREPARATION_PEDAGOGIQUE_NON_DEPLOYEE",
        "N_CHOIX_APPRENANT_DOCUMENTES": int((bloc["STATUT_PREPARATION"] == "CHOIX_APPRENANT_DOCUMENTE").sum())
    })
groupes_t0_proposes = pd.DataFrame(lignes_groupes_t0)
assert groupes_t0_proposes["ID_GROUPE"].is_unique
assert int(groupes_t0_proposes["N_APPRENANTS"].sum()) == len(valides_t0_cycles)
# L’export automatique pur et la consolidation finale sont produits dans la cellule suivante.
groupes_t0_proposes.to_csv(CYCLES_DIR / "recommandations_contenu_groupes_T0.csv", index=False, encoding="utf-8-sig")
print("T0 :", len(reco_t0_cycles), "propositions ;", len(valides_t0_cycles), "avec parcours préparé ;",
      int(reco_t0_cycles["PARCOURS_PREPARE"].isna().sum()), "indécises ;", len(groupes_t0_proposes), "groupes provisoires.")
display(groupes_t0_proposes)
```

### Cellule 24C — Quatre résultats pérennes et séparation des décisions

Ce contrôle garantit les **quatre résultats** à conserver pour l'architecture à deux cycles. La sortie automatique n'intègre aucun jugement humain. Le fichier `decisions_parcours_enseignant_apprenant_T0.csv` reste une saisie manuelle indépendante et ne sera jamais remplacé par une proposition algorithmique. Le parcours final est renseigné seulement si le **choix libre de l'apprenant** a été enregistré ; T1 sert uniquement à un audit rétrospectif du parcours observé.


```python
# ============================================================
# CELLULE 24C — Conservation des 4 résultats du parcours pédagogique
# T0 -> proposition automatique -> avis formateur -> choix apprenant -> T1
# ============================================================

# 1. Garantir la présence des deux évaluations, sans recalcul ni imputation.
F_RESULTAT_T0 = INPUT_02 / "resultat_T0_deux_cycles.csv"
F_RESULTAT_T1 = INPUT_03 / "resultat_T1_deux_cycles.csv"
if not F_RESULTAT_T0.is_file() or not F_RESULTAT_T1.is_file():
    raise FileNotFoundError(
        "Exécuter intégralement les notebooks 02 puis 03 avant le notebook 05. "
        "Le résultat T0 ou T1 consolidé est absent."
    )
RESULTAT_T0_STABLE = pd.read_csv(F_RESULTAT_T0, dtype={ID: "string"})
RESULTAT_T1_STABLE = pd.read_csv(F_RESULTAT_T1, dtype={ID: "string"})
assert len(RESULTAT_T0_STABLE) == 305 and RESULTAT_T0_STABLE[ID].is_unique
assert len(RESULTAT_T1_STABLE) == 196 and RESULTAT_T1_STABLE[ID].is_unique
assert RESULTAT_T0_STABLE[ID].notna().all() and RESULTAT_T1_STABLE[ID].notna().all()
assert set(RESULTAT_T0_STABLE[ID]) == set(BASE_RECO_MULTI[ID])
assert len(set(RESULTAT_T1_STABLE[ID]) & set(BASE_MT1_125[ID])) == 125

# 2. ARCHIVE AUTOMATIQUE : uniquement des champs dérivés des données T0 et D1.
# Ne jamais y introduire les décisions formateur ou les choix de l'apprenant.
champs_auto = [ID, "PARCOURS_PRINCIPAL", "PARCOURS_CANDIDATS", "CLASSEMENT_PARCOURS",
               "NB_PREUVES_MAX", "NB_CANDIDATS_MAX"]
if "PARCOURS_CANDIDATS_T0_BASE" in BASE_RECO_MULTI.columns:
    champs_auto.append("PARCOURS_CANDIDATS_T0_BASE")
champs_auto += [f"S_{p}" for p in PARCOURS_CYCLES]
champs_auto += [f"PREUVES_{p}" for p in PARCOURS_CYCLES]
champs_auto += [f"RAISONS_{p}" for p in PARCOURS_CYCLES]
assert set(champs_auto).issubset(BASE_RECO_MULTI.columns)
RESULTAT_RECO_AUTO = BASE_RECO_MULTI[champs_auto].copy()
RESULTAT_RECO_AUTO["PARCOURS_AUTO"] = RESULTAT_RECO_AUTO["PARCOURS_PRINCIPAL"].where(
    RESULTAT_RECO_AUTO["PARCOURS_PRINCIPAL"].isin(PARCOURS_CYCLES), pd.NA
)
RESULTAT_RECO_AUTO["STATUT_AUTO"] = np.where(
    RESULTAT_RECO_AUTO["PARCOURS_AUTO"].notna(), "PROPOSITION_UNIVOQUE", "A_VALIDER"
)
RESULTAT_RECO_AUTO["SCORE_T0_PARCOURS_AUTO"] = np.select(
    [RESULTAT_RECO_AUTO["PARCOURS_AUTO"].eq(p).fillna(False).to_numpy() for p in PARCOURS_CYCLES],
    [RESULTAT_RECO_AUTO[f"S_{p}"].to_numpy() for p in PARCOURS_CYCLES], default=np.nan
)
RESULTAT_RECO_AUTO["NIVEAU_T0_AUTO"] = RESULTAT_RECO_AUTO[
    "SCORE_T0_PARCOURS_AUTO"].map(classer_score_t0)
RESULTAT_RECO_AUTO["ETAPE_DECISION"] = "T0"
assert len(RESULTAT_RECO_AUTO) == 305 and RESULTAT_RECO_AUTO[ID].is_unique
assert RESULTAT_RECO_AUTO.columns.is_unique, "En-têtes dupliqués dans l'archive automatique"
F_RECO_AUTO = CYCLES_DIR / "resultat_recommandation_auto_T0.csv"
RESULTAT_RECO_AUTO.to_csv(F_RECO_AUTO, index=False, encoding="utf-8-sig")

# 3. ARCHIVE FINALE : proposition, avis enseignant et choix apprenant sont DISTINCTS.
# « Final » signifie choix explicite de l'apprenant ; ne pas déduire le choix du T1.
RESULTAT_RECO_FINALE = RESULTAT_RECO_AUTO[
    [ID, "PARCOURS_AUTO", "STATUT_AUTO", "NIVEAU_T0_AUTO"]].merge(
    decisions[colonnes_decision], on=ID, how="left", validate="one_to_one"
)
for c in colonnes_decision[1:]:
    RESULTAT_RECO_FINALE[c] = RESULTAT_RECO_FINALE[c].fillna("").astype(str).str.strip()
decision_normalisee = RESULTAT_RECO_FINALE["DECISION_ENSEIGNANT"].str.upper()
# La date de notification doit être une observation réelle, jamais imputée.
# Cette politique décrit un mécanisme asynchrone ; les CSV historiques sans
# notification restent « NOTIFICATION_NON_DOCUMENTEE ».
RESULTAT_RECO_FINALE["DATE_ECHEANCE_ENSEIGNANT"] = reco_t0_cycles.set_index(ID).loc[
    RESULTAT_RECO_FINALE[ID], "DATE_ECHEANCE_ENSEIGNANT"].to_numpy()
RESULTAT_RECO_FINALE["STATUT_REVUE_ENSEIGNANT"] = reco_t0_cycles.set_index(ID).loc[
    RESULTAT_RECO_FINALE[ID], "STATUT_REVUE_ENSEIGNANT"].to_numpy()

# Comparaisons sûres : pd.NA (cas A_VALIDER) est conservé dans les données.
# Les chaînes de comparaison sont temporaires et ne modifient jamais PARCOURS_AUTO.
_auto_compare = RESULTAT_RECO_FINALE["PARCOURS_AUTO"].fillna("").astype(str).str.strip()
_avis_compare = RESULTAT_RECO_FINALE["PARCOURS_ENSEIGNANT"].fillna("").astype(str).str.strip()
_auto_ok = _auto_compare.isin(PARCOURS_CYCLES)
_avis_ok = _avis_compare.isin(PARCOURS_CYCLES)
assert not _auto_compare.isna().any() and not _avis_compare.isna().any()
_trace_ok = (RESULTAT_RECO_FINALE["VALIDATEUR"].ne("")
             & RESULTAT_RECO_FINALE["DATE_VALIDATION"].ne(""))
_validation_ok = (decision_normalisee.eq("VALIDE") & _auto_ok
                  & (~_avis_ok | _avis_compare.eq(_auto_compare)))
_modification_ok = (decision_normalisee.eq("MODIFIE") & _auto_ok & _avis_ok
                    & _avis_compare.ne(_auto_compare))
_arbitrage_ok = decision_normalisee.isin(["ARBITRE", "MODIFIE"]) & ~_auto_ok & _avis_ok
_rejet_ok = decision_normalisee.eq("REJETE") & _auto_ok
_avis_date_valide = pd.to_datetime(RESULTAT_RECO_FINALE["DATE_VALIDATION"], errors="coerce", utc=True).notna()
_trace_ok = _trace_ok & _avis_date_valide
_echeance_dt = pd.to_datetime(RESULTAT_RECO_FINALE["DATE_ECHEANCE_ENSEIGNANT"], errors="coerce", utc=True)
_date_avis_dt = pd.to_datetime(RESULTAT_RECO_FINALE["DATE_VALIDATION"], errors="coerce", utc=True)
_resultat_tardif = _echeance_dt.notna() & _date_avis_dt.gt(_echeance_dt)
RESULTAT_RECO_FINALE["AVIS_ENSEIGNANT_TRACABLE"] = (
    _trace_ok & (_validation_ok | _modification_ok | _arbitrage_ok | _rejet_ok)
)
RESULTAT_RECO_FINALE["TYPE_AVIS_FORMATEUR"] = np.select(
    [_trace_ok & _validation_ok, _trace_ok & _modification_ok,
     _trace_ok & _arbitrage_ok, _trace_ok & _rejet_ok],
    ["ACCEPTATION_PROPOSITION", "MODIFICATION_PROPOSITION",
     "ARBITRAGE_CAS_AMBIGU", "REJET_PROPOSITION"],
    default="AUCUN_AVIS_EXPLOITABLE")
RESULTAT_RECO_FINALE["PARCOURS_AVIS_FORMATEUR"] = np.select(
    [RESULTAT_RECO_FINALE["AVIS_ENSEIGNANT_TRACABLE"] & decision_normalisee.eq("VALIDE"),
     RESULTAT_RECO_FINALE["AVIS_ENSEIGNANT_TRACABLE"] & decision_normalisee.isin(["MODIFIE", "ARBITRE"])
      & RESULTAT_RECO_FINALE["PARCOURS_ENSEIGNANT"].isin(PARCOURS_CYCLES)],
    [RESULTAT_RECO_FINALE["PARCOURS_AUTO"].fillna(""),
     RESULTAT_RECO_FINALE["PARCOURS_ENSEIGNANT"]], default=""
)
# Un conseil reçu tardivement est archivé, sans écraser un choix déjà documenté.
RESULTAT_RECO_FINALE["AVIS_ENSEIGNANT_TARDIF"] = (
    RESULTAT_RECO_FINALE["AVIS_ENSEIGNANT_TRACABLE"] & _resultat_tardif)
RESULTAT_RECO_FINALE["CHOIX_APPRENANT_DOCUMENTE"] = RESULTAT_RECO_FINALE[
    "PARCOURS_CHOISI"].isin(PARCOURS_CYCLES)
RESULTAT_RECO_FINALE["PARCOURS_FINAL"] = RESULTAT_RECO_FINALE[
    "PARCOURS_CHOISI"].where(RESULTAT_RECO_FINALE["CHOIX_APPRENANT_DOCUMENTE"], pd.NA)
RESULTAT_RECO_FINALE["STATUT_FINAL"] = np.where(
    RESULTAT_RECO_FINALE["CHOIX_APPRENANT_DOCUMENTE"],
    "CHOIX_LIBRE_APPRENANT_DOCUMENTE", "EN_ATTENTE_CHOIX_APPRENANT"
)
RESULTAT_RECO_FINALE["ETAPE_DECISION"] = "T0"
assert RESULTAT_RECO_FINALE[ID].is_unique and len(RESULTAT_RECO_FINALE) == 305
F_RECO_FINALE = CYCLES_DIR / "resultat_recommandation_finale_T0.csv"
# Fichier calculé régénérable. Le fichier manuel 'decisions_*' n'est jamais réécrit ici.
RESULTAT_RECO_FINALE.to_csv(F_RECO_FINALE, index=False, encoding="utf-8-sig")

# 4. AUDIT A POSTERIORI : le parcours observé à T1 est une preuve distincte
# du choix éventuellement saisi à T0 ; il ne doit jamais le remplacer.
assert "PARCOURS_T1" in RESULTAT_T1_STABLE.columns
SUIVI_T1 = RESULTAT_T1_STABLE[[ID, "PARCOURS_T1", "SCORE_T1"]].copy()
AUDIT_CHOIX_VS_T1 = RESULTAT_RECO_FINALE[
    [ID, "PARCOURS_AUTO", "PARCOURS_AVIS_FORMATEUR", "PARCOURS_FINAL",
     "STATUT_FINAL"]].merge(SUIVI_T1, on=ID, how="left", validate="one_to_one")
choix_compare = AUDIT_CHOIX_VS_T1["PARCOURS_FINAL"].fillna("").astype(str)
suivi_compare = AUDIT_CHOIX_VS_T1["PARCOURS_T1"].fillna("").astype(str)
comparaison_possible = choix_compare.ne("") & suivi_compare.ne("")
AUDIT_CHOIX_VS_T1["CONCORDANCE_CHOIX_ET_T1"] = np.select(
    [comparaison_possible & choix_compare.eq(suivi_compare),
     comparaison_possible & choix_compare.ne(suivi_compare)],
    ["CONCORDANT", "DIVERGENT"], default="NON_EVALUABLE"
)
AUDIT_CHOIX_VS_T1.to_csv(CYCLES_DIR / "audit_choix_apprenant_vs_parcours_suivi_T1.csv",
                          index=False, encoding="utf-8-sig")

print("\nQUATRE RÉSULTATS PÉRENNISÉS")
print("1. T0 diagnostiqué :", len(RESULTAT_T0_STABLE), F_RESULTAT_T0)
print("2. Recommandations automatiques :", len(RESULTAT_RECO_AUTO), F_RECO_AUTO)
print("3. Recommandations finales :", len(RESULTAT_RECO_FINALE), F_RECO_FINALE,
      "| Choix d'apprenants documentés :",
      int(RESULTAT_RECO_FINALE["CHOIX_APPRENANT_DOCUMENTE"].sum()))
print("4. T1 évalué :", len(RESULTAT_T1_STABLE), F_RESULTAT_T1)
print("Journal MANUEL préservé :", f_decision)
```

### H4 — Vérification de l'acceptabilité pédagogique à T0

Le taux d'acceptation est calculé **uniquement** à partir des validations humaines datées et attribuées à un enseignant. S'il n'existe pas de validation, H4 reste *non évaluable* ; la concordance avec le parcours finalement suivi ne remplace pas l'avis pédagogique de l'enseignant.

H4 distingue les jugements traçables, les arbitrages de recommandations ambiguës, les expirations sans avis et les avis tardifs. Le statut `EXPIRE_SANS_AVIS` ne vaut jamais `VALIDE`.


```python
# ============================================================
# H4 — Avis enseignants TRAÇABLES, distincts des arbitrages
# En absence de décision humaine : H4 reste NON_EVALUABLE.
# ============================================================

import pandas as pd

assert reco_t0_cycles[ID].is_unique and len(reco_t0_cycles) == 305

decisions_h4 = reco_t0_cycles["DECISION_ENSEIGNANT"].fillna("").astype(str).str.upper().str.strip()
autorisees_h4 = {"", "VALIDE", "MODIFIE", "REJETE", "EN_ATTENTE", "ARBITRE", "EXPIRE_SANS_AVIS", "AVIS_TARDIF"}
assert set(decisions_h4) <= autorisees_h4, "Valeur DECISION_ENSEIGNANT inconnue."

# Comparer deux parcours via des copies normalisées évite TypeError sur pd.NA.
_auto_h4_compare = reco_t0_cycles["PARCOURS_AUTO"].fillna("").astype(str).str.strip()
_enseignant_h4_compare = reco_t0_cycles["PARCOURS_ENSEIGNANT"].fillna("").astype(str).str.strip()
parcours_auto_valide = _auto_h4_compare.isin(PARCOURS_CYCLES)
parcours_enseignant_valide = _enseignant_h4_compare.isin(PARCOURS_CYCLES)
trace_complete = (reco_t0_cycles["VALIDATEUR"].fillna("").astype(str).str.strip().ne("")
                  & reco_t0_cycles["DATE_VALIDATION"].fillna("").astype(str).str.strip().ne(""))
statut_jugement = decisions_h4.isin({"VALIDE", "MODIFIE", "REJETE", "ARBITRE"})
# Les avis tardifs peuvent être analysés séparément mais ne sont pas comptés
# dans le taux H4 « dans le délai » (lorsqu’une notification est documentée).
_dans_delai_h4 = ~reco_t0_cycles["STATUT_REVUE_ENSEIGNANT"].eq("AVIS_TARDIF")

# Un VALIDE n'a de sens que si une proposition unique existe ;
# MODIFIE exige une proposition unique et un autre parcours renseigné ;
# REJETE vise également une proposition unique.
acceptation_valide = trace_complete & parcours_auto_valide & decisions_h4.eq("VALIDE") & (
    ~parcours_enseignant_valide |
    _enseignant_h4_compare.eq(_auto_h4_compare)
)
modification_valide = (trace_complete & parcours_auto_valide & decisions_h4.eq("MODIFIE")
                       & parcours_enseignant_valide
                       & _enseignant_h4_compare.ne(_auto_h4_compare))
rejet_valide = trace_complete & parcours_auto_valide & decisions_h4.eq("REJETE")
masque_h4 = (acceptation_valide | modification_valide | rejet_valide) & _dans_delai_h4
acceptation_valide = acceptation_valide & _dans_delai_h4
modification_valide = modification_valide & _dans_delai_h4
rejet_valide = rejet_valide & _dans_delai_h4

# Un MODIFIE sur un cas A_VALIDER est un arbitrage : ce n'est pas une acceptation H4.
masque_arbitrage = (_dans_delai_h4 & trace_complete & ~parcours_auto_valide
                    & decisions_h4.isin(["MODIFIE", "ARBITRE"]) & parcours_enseignant_valide)
masque_incoherent = (trace_complete & statut_jugement & _dans_delai_h4
                    & ~(masque_h4 | masque_arbitrage))
masque_sans_trace = statut_jugement & ~trace_complete

mesures_h4 = reco_t0_cycles.loc[masque_h4].copy()
nb_h4 = len(mesures_h4)
synthese_h4 = pd.DataFrame([{
    "N_PROPOSITIONS": len(reco_t0_cycles),
    "N_PROPOSITIONS_UNIVOQUES": int(parcours_auto_valide.sum()),
    "N_CAS_AMBIGUS": int((~parcours_auto_valide).sum()),
    "N_JUGEMENTS_ENSEIGNANTS_TRACABLES": nb_h4,
    "N_VALIDE": int(acceptation_valide.sum()),
    "N_MODIFIE": int(modification_valide.sum()),
    "N_REJETE": int(rejet_valide.sum()),
    "N_ARBITRAGES_CAS_AMBIGUS": int(masque_arbitrage.sum()),
    "N_JUGEMENTS_INCOHERENTS": int(masque_incoherent.sum()),
    "N_JUGEMENTS_SANS_TRACE": int(masque_sans_trace.sum()),
    "N_EXPIRES_SANS_AVIS": int(reco_t0_cycles["STATUT_REVUE_ENSEIGNANT"].eq("EXPIRE_SANS_AVIS").sum()),
    "N_AVIS_TARDIFS": int(reco_t0_cycles["STATUT_REVUE_ENSEIGNANT"].eq("AVIS_TARDIF").sum()),
    "N_NOTIFICATION_NON_DOCUMENTEE": int(reco_t0_cycles["STATUT_REVUE_ENSEIGNANT"].eq("NOTIFICATION_NON_DOCUMENTEE").sum()),
    "TAUX_ACCEPTATION_PARMI_JUGES": (float(acceptation_valide.sum()/nb_h4) if nb_h4 else np.nan),
    "STATUT_H4": ("DESCRIPTIF_JUGEMENTS_OBSERVES" if nb_h4 and not masque_incoherent.any()
                  else ("EXAMEN_INCOHERENCES_REQUIS" if masque_incoherent.any()
                        else "NON_EVALUABLE_SANS_JUGEMENTS_ENSEIGNANTS"))
}])
synthese_h4.to_csv(CYCLES_DIR / "H4_avis_enseignant_parcours_T0.csv", index=False, encoding="utf-8-sig")

journal_h4 = reco_t0_cycles[[ID, "PARCOURS_AUTO", "DECISION_ENSEIGNANT",
                              "PARCOURS_ENSEIGNANT", "VALIDATEUR", "DATE_VALIDATION",
                              "DATE_NOTIFICATION_ENSEIGNANT", "STATUT_REVUE_ENSEIGNANT"]].copy()
journal_h4["STATUT_AUDIT_H4"] = np.select(
    [masque_h4, masque_arbitrage, masque_incoherent, masque_sans_trace],
    ["JUGEMENT_H4_EVALUABLE", "ARBITRAGE_AMBIGU_SEPARE", "DECISION_INCOHERENTE", "TRACE_INCOMPLETE"],
    default="AUCUNE_DECISION_HUMAINE_EVALUABLE")
journal_h4.to_csv(CYCLES_DIR / "audit_H4_jugements_T0.csv", index=False, encoding="utf-8-sig")
display(synthese_h4)
print("H4 :", synthese_h4.loc[0, "STATUT_H4"])
```

# Cellule 29 — Clustering séparé dans chaque parcours

```python
# ============================================================
# CELLULE 29 — Segmentation exploratoire OOF par parcours
# ============================================================

from sklearn.cluster import KMeans


BASE_MT1_125[
    "NIVEAU_PARCOURS"
] = pd.NA


BASE_MT1_125[
    "STATUT_NIVEAU"
] = (
    "EXPLORATOIRE_KMEANS_OOF"
)


resume_clusters_parcours = []


for parcours in [
    "DA",
    "BI",
    "DS",
    "IA"
]:

    masque = (
        BASE_MT1_125[
            "T1__PARCOURS"
        ] == parcours
    )


    temp = (
        BASE_MT1_125.loc[
            masque,
            [
                "T2_PREDIT_OOF"
            ]
        ]
        .copy()
    )


    assert len(temp) >= 3


    kmeans = KMeans(
        n_clusters=3,
        random_state=RANDOM_STATE,
        n_init=50
    )


    clusters = (
        kmeans.fit_predict(
            temp[
                ["T2_PREDIT_OOF"]
            ]
        )
    )


    centres = (
        pd.DataFrame({
            "CLUSTER":
                [0, 1, 2],

            "CENTRE":
                kmeans
                .cluster_centers_
                .ravel()
        })

        .sort_values(
            "CENTRE"
        )

        .reset_index(
            drop=True
        )
    )


    mapping = {
        int(
            centres.iloc[0][
                "CLUSTER"
            ]
        ):
            "Debutant",

        int(
            centres.iloc[1][
                "CLUSTER"
            ]
        ):
            "Intermediaire",

        int(
            centres.iloc[2][
                "CLUSTER"
            ]
        ):
            "Avance"
    }


    BASE_MT1_125.loc[
        temp.index,
        "NIVEAU_PARCOURS"
    ] = (
        pd.Series(
            clusters,
            index=temp.index
        )
        .map(mapping)
    )


    for _, ligne in (
        centres.iterrows()
    ):

        resume_clusters_parcours.append({
            "PARCOURS":
                parcours,

            "NIVEAU_EXPLORATOIRE":
                mapping[
                    int(
                        ligne[
                            "CLUSTER"
                        ]
                    )
                ],

            "CENTRE_T2_PREDIT_OOF":
                ligne["CENTRE"]
        })


CENTRES_NIVEAUX = pd.DataFrame(
    resume_clusters_parcours
)


display(
    CENTRES_NIVEAUX
)
# Ces clusters sont descriptifs et exploratoires : les prédictions OOF reposent
# sur des modèles évalués rétrospectivement (notebook 03), non sur une décision T0.
assert BASE_MT1_125["NIVEAU_PARCOURS"].notna().all()
assert CENTRES_NIVEAUX.groupby("PARCOURS").size().eq(3).all()
```

# Cellule 36 — Préparer les prérequis par parcours

```python
# ============================================================
# CELLULE 36 — Profil de prérequis
# ============================================================

def profil_prerequis(row):

    parcours = row["T1__PARCOURS"]

    if parcours == "DA":
        return (
            f"INFO={row['P__NIVEAU_INFORMATIQUE']} | "
            f"EXCEL={row['P__NIVEAU_EXCEL']}"
        )

    if parcours == "BI":
        return (
            f"INFO={row['P__NIVEAU_INFORMATIQUE']} | "
            f"EXCEL={row['P__NIVEAU_EXCEL']} | "
            f"POWERBI={row['P__NIVEAU_POWER_BI']}"
        )

    if parcours == "DS":
        return (
            f"INFO={row['P__NIVEAU_INFORMATIQUE']} | "
            f"PYTHON={row['P__NIVEAU_PYTHON']}"
        )

    if parcours == "IA":
        return (
            f"INFO={row['P__NIVEAU_INFORMATIQUE']} | "
            f"PYTHON={row['P__NIVEAU_PYTHON']} | "
            f"IA={row['P__NIVEAU_IA']}"
        )

    return "NON_DEFINI"


BASE_GROUPES[
    "PROFIL_PREREQUIS"
] = BASE_GROUPES.apply(
    profil_prerequis,
    axis=1
)


print("✅ Prérequis ajoutés")
```

# Cellule 38 — Niveau de prérequis du parcours
Groupe =Parcours+ Niveau+ Lacunes communes+ Prérequis compatibles

```python
# ============================================================
# CELLULE 38 — Niveau de prérequis limitant
# ============================================================

NIVEAU_LABEL = {
    0: "Aucun",
    1: "Debutant",
    2: "Intermediaire",
    3: "Avance"
}


def prerequis_limitant(row):

    parcours = row["T1__PARCOURS"]

    if parcours == "DA":
        valeurs = [
            row["P__NIVEAU_INFORMATIQUE"],
            row["P__NIVEAU_EXCEL"]
        ]

    elif parcours == "BI":
        valeurs = [
            row["P__NIVEAU_INFORMATIQUE"],
            row["P__NIVEAU_EXCEL"],
            row["P__NIVEAU_POWER_BI"]
        ]

    elif parcours == "DS":
        valeurs = [
            row["P__NIVEAU_INFORMATIQUE"],
            row["P__NIVEAU_PYTHON"]
        ]

    elif parcours == "IA":
        valeurs = [
            row["P__NIVEAU_INFORMATIQUE"],
            row["P__NIVEAU_PYTHON"],
            row["P__NIVEAU_IA"]
        ]

    else:
        return pd.NA

    valeurs = [
        v
        for v in valeurs
        if pd.notna(v)
    ]

    if len(valeurs) == 0:
        return pd.NA

    return NIVEAU_LABEL[
        int(min(valeurs))
    ]


BASE_GROUPES[
    "PREREQUIS_LIMITANT"
] = BASE_GROUPES.apply(
    prerequis_limitant,
    axis=1
)


display(
    BASE_GROUPES[
        [
            ID,
            "T1__PARCOURS",
            "NIVEAU_PARCOURS",
            "PREREQUIS_LIMITANT"
        ]
    ].head(10)
)
```

# Cellule 41 — Déterminer la lacune principale de chaque apprenant

Pour éviter de créer un groupe différent pour chaque combinaison de questions, on attribue à chaque apprenant sa lacune la plus commune parmi les apprenants du même parcours et niveau.

# Cellule 43 — Créer les groupes pédagogiques

```python
# ============================================================
# CELLULE 43 — Groupes pédagogiques collectifs
# ============================================================

def top_modalites(
    serie,
    n=3
):

    s = (
        serie
        .dropna()
        .astype(str)
    )

    s = s[
        ~s.str.startswith(
            "Aucune"
        )
    ]

    if len(s) == 0:
        return "Aucune"

    return " | ".join(
        s.value_counts()
        .head(n)
        .index
        .tolist()
    )


def modalite_dominante(
    serie
):

    s = (
        serie
        .dropna()
        .astype(str)
    )

    if len(s) == 0:
        return pd.NA

    return (
        s.value_counts()
        .index[0]
    )


groupes_pedagogiques = (

    BASE_GROUPES

    .groupby(
        [
            "T1__PARCOURS",
            "NIVEAU_PARCOURS"
        ],
        as_index=False,
        dropna=False
    )

    .agg(
        N_APPRENANTS=(
            ID,
            "nunique"
        ),

        T1_MOYEN=(
            "T1__SCORE",
            "mean"
        ),

        T2_PREDIT_OOF_MOYEN=(
            "T2_PREDIT_OOF",
            "mean"
        ),

        NB_LACUNES_MOYEN=(
            "NB_LACUNES",
            "mean"
        ),

        LACUNES_PRIORITAIRES=(
            "LACUNE_PRINCIPALE",
            top_modalites
        ),

        PREREQUIS_DOMINANT=(
            "PREREQUIS_LIMITANT",
            modalite_dominante
        ),

        OBJECTIFS_DOMINANTS=(
            "Objectif pro",
            lambda x:
                top_modalites(
                    x,
                    n=2
                )
        )
    )
)


groupes_pedagogiques = (
    groupes_pedagogiques
    .sort_values(
        [
            "T1__PARCOURS",
            "NIVEAU_PARCOURS"
        ]
    )
    .reset_index(drop=True)
)


groupes_pedagogiques[
    "ID_GROUPE"
] = [
    f"G{n:03d}"
    for n in range(
        1,
        len(groupes_pedagogiques) + 1
    )
]


display(
    groupes_pedagogiques
)
```

# Cellule 50 — Remédiation spécifique aux lacunes

```python
# ============================================================
# CELLULE 50 — Remédiation selon les lacunes du groupe
# ============================================================

# ------------------------------------------------------------
# 1. Vérifications
# ------------------------------------------------------------

colonnes_requises = {
    "ID_GROUPE",
    "LACUNES_PRIORITAIRES"
}

colonnes_manquantes = (
    colonnes_requises
    -
    set(BASE_GROUPES.columns)
)

assert len(colonnes_manquantes) == 0, (
    "Colonnes manquantes dans BASE_GROUPES : "
    f"{colonnes_manquantes}. "
    "Exécuter d'abord les cellules 42 à 47 corrigées."
)


# ------------------------------------------------------------
# 2. Fonction de remédiation par groupe
# ------------------------------------------------------------

def construire_remediation(row):

    lacunes = row[
        "LACUNES_PRIORITAIRES"
    ]

    # --------------------------------------------------------
    # Aucun besoin ciblé détecté
    # --------------------------------------------------------

    if pd.isna(lacunes):

        return (
            "Aucune lacune prioritaire identifiée au niveau du groupe ; "
            "poursuivre le programme commun du parcours."
        )


    lacunes = str(
        lacunes
    ).strip()


    if (
        lacunes == ""
        or lacunes.lower() == "nan"
        or lacunes in {
            "Aucune",
            "Aucune_lacune_T1",
            "Aucune lacune T1 détectée"
        }
    ):

        return (
            "Aucune lacune prioritaire identifiée au niveau du groupe ; "
            "poursuivre le programme commun du parcours."
        )


    # --------------------------------------------------------
    # Lacunes communes / prioritaires identifiées
    # --------------------------------------------------------

    return (
        "Renforcer prioritairement les compétences suivantes : "
        f"{lacunes}. "
        "Sélectionner dans le programme commun des ressources "
        "de renforcement correspondantes."
    )


# ------------------------------------------------------------
# 3. Application
# ------------------------------------------------------------

BASE_GROUPES[
    "REMEDIATION_LACUNES"
] = (
    BASE_GROUPES
    .apply(
        construire_remediation,
        axis=1
    )
)


# ------------------------------------------------------------
# 4. Vérification :
#    une même remédiation doit être associée à un même groupe
# ------------------------------------------------------------

controle_remediation = (
    BASE_GROUPES
    .groupby(
        "ID_GROUPE"
    )
    ["REMEDIATION_LACUNES"]
    .nunique()
)


assert (
    controle_remediation <= 1
).all(), (
    "Un même groupe possède plusieurs recommandations "
    "de remédiation différentes."
)


# ------------------------------------------------------------
# 5. Affichage synthétique
# ------------------------------------------------------------

resume_remediation = (
    BASE_GROUPES[
        [
            "ID_GROUPE",
            "T1__PARCOURS",
            "NIVEAU_PARCOURS",
            "LACUNES_PRIORITAIRES",
            "REMEDIATION_LACUNES"
        ]
    ]
    .drop_duplicates(
        subset=[
            "ID_GROUPE"
        ]
    )
    .sort_values(
        "ID_GROUPE"
    )
    .reset_index(
        drop=True
    )
)


display(
    resume_remediation
)


print(
    "✅ Remédiation construite pour",
    resume_remediation[
        "ID_GROUPE"
    ].nunique(),
    "groupes."
)
```

```python
# ============================================================
# CELLULE 51 — Prérequis
# ============================================================

def adaptation_prerequis(prerequis):

    if pd.isna(prerequis):
        return "Prerequis a verifier"

    if prerequis == "Aucun":
        return "Remediation prerequis prioritaire"

    if prerequis == "Debutant":
        return "Renforcement des prerequis"

    if prerequis == "Intermediaire":
        return "Rappel cible si necessaire"

    if prerequis == "Avance":
        return "Pas de renforcement prioritaire"

    return "Prerequis a verifier"


BASE_GROUPES[
    "ADAPTATION_PREREQUIS"
] = (
    BASE_GROUPES[
        "PREREQUIS_DOMINANT"
    ]
    .apply(
        adaptation_prerequis
    )
)
```

# Cellule 53 — Requête RAG corrigée
Le RAG doit chercher dans le programme du parcours, pas inventer un autre programme.

```python
# ============================================================
# CELLULE 53 — Requête RAG par groupe
# ============================================================

# ------------------------------------------------------------
# 1. Vérification des colonnes nécessaires
# ------------------------------------------------------------

colonnes_requises = {
    "ID_GROUPE",
    "PARCOURS",
    "PROGRAMME_COMMUN",
    "NIVEAU",
    "STATUT_NIVEAU",
    "N_APPRENANTS",
    "LACUNE_PRINCIPALE",
    "PREREQUIS_LIMITANT",
    "OBJECTIFS_DOMINANTS",
    "TYPE_ACCOMPAGNEMENT",
    "EXPLICATION",
    "EXERCICES",
    "REMEDIATION_LACUNES",
    "ADAPTATION_PREREQUIS",
    "SOURCE_LACUNES"
}


colonnes_manquantes = (
    colonnes_requises
    -
    set(RECOMMANDATIONS_CONTENU.columns)
)


assert len(colonnes_manquantes) == 0, (
    "Colonnes manquantes dans RECOMMANDATIONS_CONTENU : "
    f"{colonnes_manquantes}. "
    "Exécuter d'abord la Cellule 52 corrigée."
)


# ------------------------------------------------------------
# 2. Fonction de nettoyage pour éviter 'nan' dans le texte
# ------------------------------------------------------------

def texte_valide(valeur, defaut="Non renseigné"):

    if pd.isna(valeur):
        return defaut

    texte = str(valeur).strip()

    if (
        texte == ""
        or texte.lower() == "nan"
    ):
        return defaut

    return texte


# ------------------------------------------------------------
# 3. Construction de la requête RAG
# ------------------------------------------------------------

def construire_requete_rag(row):

    programme = texte_valide(
        row["PROGRAMME_COMMUN"]
    )

    parcours = texte_valide(
        row["PARCOURS"]
    )

    niveau = texte_valide(
        row["NIVEAU"]
    )

    statut_niveau = texte_valide(
        row["STATUT_NIVEAU"]
    )

    lacunes = texte_valide(
        row["LACUNE_PRINCIPALE"],
        defaut="Aucune lacune prioritaire identifiée"
    )

    prerequis = texte_valide(
        row["PREREQUIS_LIMITANT"],
        defaut="Aucun prérequis limitant identifié"
    )

    objectifs = texte_valide(
        row["OBJECTIFS_DOMINANTS"],
        defaut="Objectifs non renseignés"
    )

    accompagnement = texte_valide(
        row["TYPE_ACCOMPAGNEMENT"]
    )

    explication = texte_valide(
        row["EXPLICATION"]
    )

    exercices = texte_valide(
        row["EXERCICES"]
    )

    remediation = texte_valide(
        row["REMEDIATION_LACUNES"]
    )

    adaptation_prerequis = texte_valide(
        row["ADAPTATION_PREREQUIS"]
    )

    source_lacunes = texte_valide(
        row["SOURCE_LACUNES"],
        defaut="T1"
    )


    return (
        f"Programme commun : {programme}. "
        f"Parcours : {parcours}. "
        f"Niveau exploratoire du groupe : {niveau}. "
        f"Statut du niveau : {statut_niveau}. "
        f"Conserver strictement le même programme, "
        f"les mêmes chapitres et les mêmes objectifs pédagogiques "
        f"du parcours. "
        f"Ne pas créer un nouveau programme. "
        f"Adapter uniquement l'accompagnement pédagogique. "
        f"Lacunes prioritaires du groupe : {lacunes}. "
        f"Source des lacunes : {source_lacunes}. "
        f"Prérequis limitant dominant : {prerequis}. "
        f"Objectifs dominants du groupe : {objectifs}. "
        f"Type d'accompagnement : {accompagnement}. "
        f"Explications attendues : {explication}. "
        f"Exercices attendus : {exercices}. "
        f"Remédiation : {remediation}. "
        f"Adaptation des prérequis : {adaptation_prerequis}."
    )


# ------------------------------------------------------------
# 4. Construire la requête pour chaque groupe
# ------------------------------------------------------------

RECOMMANDATIONS_CONTENU[
    "REQUETE_RAG"
] = (
    RECOMMANDATIONS_CONTENU
    .apply(
        construire_requete_rag,
        axis=1
    )
)


# ------------------------------------------------------------
# 5. Contrôles
# ------------------------------------------------------------

assert (
    RECOMMANDATIONS_CONTENU[
        "REQUETE_RAG"
    ]
    .notna()
    .all()
)


assert (
    RECOMMANDATIONS_CONTENU[
        "REQUETE_RAG"
    ]
    .str.len()
    > 0
).all()


# ------------------------------------------------------------
# 6. Affichage
# ------------------------------------------------------------

display(
    RECOMMANDATIONS_CONTENU[
        [
            "ID_GROUPE",
            "PARCOURS",
            "PROGRAMME_COMMUN",
            "NIVEAU",
            "LACUNE_PRINCIPALE",
            "PREREQUIS_LIMITANT",
            "OBJECTIFS_DOMINANTS",
            "REQUETE_RAG"
        ]
    ]
    .head(10)
)


print(
    "✅ Requêtes RAG construites pour",
    len(RECOMMANDATIONS_CONTENU),
    "groupes."
)
```
