# Cellule 1 — Imports

```python
# ============================================================
# ÉTAPE 4 — RECOMMANDATION INITIALE DE PARCOURS
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
import json
from datetime import datetime

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 100)
```

# Cellule 2 — Chemins et chargement

```python
# ============================================================
# 4.1. Chargement du dataset d'orientation scoré
# ============================================================

INPUT_DIR = Path("outputs/02_scoring_pedagogique")
OUTPUT_DIR = Path("outputs/04_recommandation_parcours")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PATH_ORIENTATION_SCORE = INPUT_DIR / "04_dataset_orientation_score.csv"

df_orientation = pd.read_csv(PATH_ORIENTATION_SCORE)

print("Dataset orientation chargé :", df_orientation.shape)
display(df_orientation.head())
```

# Cellule 3 — Paramètres de recommandation

```python
# ============================================================
# 4.2. Paramètres pédagogiques
# ============================================================

SEUIL_DEBUTANT = 50
SEUIL_INTERMEDIAIRE = 70

domaines = ["DA", "BI", "DS", "IA"]

ordre_parcours = ["DA", "BI", "DS", "IA"]

description_parcours = {
    "DA": "Data Analysis / Excel / bases de l'analyse",
    "BI": "Business Intelligence / Power BI",
    "DS": "Data Science / Python / Machine Learning",
    "IA": "Intelligence Artificielle / LLM / RAG / Agents IA"
}
```

# Cellule 4 — Fonction de recommandation

```python
# ============================================================
# 4.3. Fonction de recommandation initiale
# ============================================================

def recommander_parcours(row):
    score_da = row["score_initial_DA_pct"]
    score_bi = row["score_initial_BI_pct"]
    score_ds = row["score_initial_DS_pct"]
    score_ia = row["score_initial_IA_pct"]
    score_global = row["score_initial_pourcentage"]
    
    scores = {
        "DA": score_da,
        "BI": score_bi,
        "DS": score_ds,
        "IA": score_ia
    }
    
    domaine_plus_faible = min(scores, key=scores.get)
    
    lacunes = [
        domaine for domaine, score in scores.items()
        if score < SEUIL_DEBUTANT
    ]
    
    # Règle pédagogique progressive
    if score_global < 30:
        parcours = "DA"
        raison = "Niveau global très faible : consolidation des bases recommandée."
    elif score_da < SEUIL_DEBUTANT:
        parcours = "DA"
        raison = "Lacunes en bases d'analyse de données."
    elif score_bi < SEUIL_DEBUTANT:
        parcours = "BI"
        raison = "Bases DA suffisantes, mais lacunes en Business Intelligence."
    elif score_ds < SEUIL_DEBUTANT:
        parcours = "DS"
        raison = "Bases DA/BI suffisantes, mais lacunes en Data Science."
    elif score_ia < SEUIL_DEBUTANT:
        parcours = "IA"
        raison = "Bases suffisantes, mais lacunes en Intelligence Artificielle."
    else:
        parcours = domaine_plus_faible
        raison = "Aucune lacune critique ; recommandation basée sur le domaine le plus faible."
    
    return pd.Series({
        "parcours_recommande": parcours,
        "description_parcours_recommande": description_parcours[parcours],
        "domaine_plus_faible": domaine_plus_faible,
        "lacunes_detectees": ", ".join(lacunes) if lacunes else "Aucune lacune critique",
        "raison_recommandation": raison
    })
```

# Cellule 5 — Application de la recommandation

```python
# ============================================================
# 4.4. Application de la recommandation
# ============================================================

df_recommandations = df_orientation.copy()

resultats_reco = df_recommandations.apply(
    recommander_parcours,
    axis=1
)

df_recommandations = pd.concat(
    [df_recommandations, resultats_reco],
    axis=1
)

print("Recommandations générées :", df_recommandations.shape)

print("\nRépartition des parcours recommandés :")
display(df_recommandations["parcours_recommande"].value_counts())

print("\nAperçu des recommandations :")
display(df_recommandations[
    [
        "IDENTIFICATION",
        "score_initial_total",
        "score_initial_pourcentage",
        "niveau_initial_global",
        "score_initial_DA_pct",
        "score_initial_BI_pct",
        "score_initial_DS_pct",
        "score_initial_IA_pct",
        "parcours_recommande",
        "domaine_plus_faible",
        "lacunes_detectees",
        "raison_recommandation"
    ]
].head(20))
```

# Cellule 7 — Comparaison entre parcours recommandé et parcours réellement suivi

Cette cellule est créer la variable importante :

Elle servira plus tard pour comparer :
apprenants ayant appliqué la recommandation IA

```python
# ============================================================
# 4.6. Comparaison entre parcours recommandé et parcours réellement suivi
# ============================================================

# Chargement du dataset prédiction scoré si nécessaire
PATH_PREDICTION_SCORE = Path("outputs/02_scoring_pedagogique/05_dataset_prediction_score.csv")

if "df_prediction_score" not in globals():
    df_prediction_score = pd.read_csv(PATH_PREDICTION_SCORE)

# Colonnes utiles du parcours réellement suivi
colonnes_suivi = [
    "IDENTIFICATION",
    "parcours_test_final",
    "score_final_total",
    "score_final_pourcentage",
    "niveau_final",
    "progression_absolue",
    "reussite_finale"
]

df_suivi = df_prediction_score[colonnes_suivi].copy()

# Fusion recommandation + parcours réellement suivi
df_comparaison_reco_suivi = df_recommandations.merge(
    df_suivi,
    on="IDENTIFICATION",
    how="inner",
    validate="one_to_one"
)

# Respect de la recommandation
df_comparaison_reco_suivi["respect_recommandation"] = (
    df_comparaison_reco_suivi["parcours_recommande"]
    == df_comparaison_reco_suivi["parcours_test_final"]
)

df_comparaison_reco_suivi["groupe_recommandation"] = np.where(
    df_comparaison_reco_suivi["respect_recommandation"],
    "A suivi la recommandation IA",
    "A suivi son propre choix"
)

print("Comparaison recommandation / parcours suivi :", df_comparaison_reco_suivi.shape)

print("\nRespect de la recommandation :")
display(df_comparaison_reco_suivi["respect_recommandation"].value_counts())

print("\nRépartition des groupes :")
display(df_comparaison_reco_suivi["groupe_recommandation"].value_counts())

print("\nMatrice parcours recommandé / parcours suivi :")
matrice_reco_suivi = pd.crosstab(
    df_comparaison_reco_suivi["parcours_recommande"],
    df_comparaison_reco_suivi["parcours_test_final"]
).reindex(index=ordre_parcours, columns=ordre_parcours, fill_value=0)

display(matrice_reco_suivi)

# Export
PATH_COMPARAISON_RECO_SUIVI = OUTPUT_DIR / "03_comparaison_recommandation_parcours_suivi.csv"
PATH_MATRICE_RECO_SUIVI = OUTPUT_DIR / "04_matrice_recommandation_parcours_suivi.csv"

df_comparaison_reco_suivi.to_csv(
    PATH_COMPARAISON_RECO_SUIVI,
    index=False,
    encoding="utf-8-sig"
)

matrice_reco_suivi.to_csv(
    PATH_MATRICE_RECO_SUIVI,
    encoding="utf-8-sig"
)

print("\nFichier comparaison exporté :", PATH_COMPARAISON_RECO_SUIVI)
print("Matrice exportée :", PATH_MATRICE_RECO_SUIVI)

display(df_comparaison_reco_suivi[
    [
        "IDENTIFICATION",
        "score_initial_total",
        "niveau_initial_global",
        "parcours_recommande",
        "parcours_test_final",
        "respect_recommandation",
        "groupe_recommandation",
        "score_final_pourcentage",
        "niveau_final",
        "progression_absolue"
    ]
].head(20))
```

# Cellule 9 — Test statistique recommandation IA vs choix personnel

```python
# ============================================================
# 4.8. Tests statistiques : recommandation IA vs choix personnel
# ============================================================

from scipy import stats
import numpy as np
import pandas as pd

df_groupe_ia = df_comparaison_reco_suivi[
    df_comparaison_reco_suivi["respect_recommandation"] == True
].copy()

df_groupe_choix = df_comparaison_reco_suivi[
    df_comparaison_reco_suivi["respect_recommandation"] == False
].copy()


def cohen_d(x, y):
    x = pd.Series(x).dropna()
    y = pd.Series(y).dropna()
    
    nx = len(x)
    ny = len(y)
    
    if nx < 2 or ny < 2:
        return np.nan
    
    pooled_std = np.sqrt(
        ((nx - 1) * x.var(ddof=1) + (ny - 1) * y.var(ddof=1))
        / (nx + ny - 2)
    )
    
    if pooled_std == 0:
        return np.nan
    
    return (x.mean() - y.mean()) / pooled_std


variables_a_tester = [
    ("score_final_pourcentage", "Score final (%)"),
    ("progression_absolue", "Progression initiale-finale")
]

resultats_tests = []

for colonne, label in variables_a_tester:
    x = df_groupe_ia[colonne].dropna()
    y = df_groupe_choix[colonne].dropna()
    
    t_stat, p_ttest = stats.ttest_ind(x, y, equal_var=False)
    u_stat, p_mann = stats.mannwhitneyu(x, y, alternative="two-sided")
    
    resultats_tests.append({
        "variable": label,
        "n_recommandation_ia": len(x),
        "n_choix_personnel": len(y),
        "moyenne_recommandation_ia": round(x.mean(), 2),
        "moyenne_choix_personnel": round(y.mean(), 2),
        "difference_moyenne": round(x.mean() - y.mean(), 2),
        "p_value_ttest_welch": round(p_ttest, 4),
        "p_value_mann_whitney": round(p_mann, 4),
        "cohen_d": round(cohen_d(x, y), 4)
    })

df_tests_recommandation = pd.DataFrame(resultats_tests)

print("===== Tests statistiques : recommandation IA vs choix personnel =====")
display(df_tests_recommandation)

# Test du taux de réussite
table_reussite = pd.crosstab(
    df_comparaison_reco_suivi["groupe_recommandation"],
    df_comparaison_reco_suivi["reussite_finale"]
)

print("\n===== Tableau réussite finale =====")
display(table_reussite)

chi2, p_chi2, dof, expected = stats.chi2_contingency(table_reussite)

print("\nTest du Khi-deux sur la réussite finale")
print("chi2 :", round(chi2, 4))
print("p-value :", round(p_chi2, 4))

# Export
PATH_TESTS_RECOMMANDATION = OUTPUT_DIR / "06_tests_statistiques_recommandation.csv"

df_tests_recommandation.to_csv(
    PATH_TESTS_RECOMMANDATION,
    index=False,
    encoding="utf-8-sig"
)

print("\nFichier exporté :", PATH_TESTS_RECOMMANDATION)
```

#  Cellule 11 — Recommandation de contenus par parcours et niveau

La recommandation de contenu vient juste après la recommandation de parcours.

Étape 4 — Recommandation initiale de parcours
Étape 4 bis — Recommandation de contenus par parcours et par niveau
Étape 4 ter — Ajustement des contenus après le test intermédiaire

Pourquoi ?

Parce que pour recommander un contenu, il faut connaître :

1. le parcours recommandé : DA / BI / DS / IA ;
2. le niveau ou cluster : Débutant / Intermédiaire / Avancé ;
3. les lacunes détectées dans le test initial ;
4. les lacunes détectées dans le test intermédiaire.

La logique devient donc :

Test initial
→ scores par domaine
→ niveau / cluster
→ parcours recommandé
→ contenus recommandés au début du parcours

Puis :

Test intermédiaire
→ détection des lacunes restantes
→ ajustement ou remédiation des contenus
→ exercices ciblés avant le test final

Exemple :

Parcours recommandé : DA
Niveau : Débutant
Lacunes initiales : Excel, nettoyage des données, TCD
Contenus recommandés :
- bases Excel ;
- nettoyage de données ;
- tableaux croisés dynamiques ;
- exercices simples.

Après test intermédiaire :
Lacune détectée : jointure ou modèle de données
Contenus de remédiation :
- rappel sur les clés de jointure ;
- exercices sur Power Query ;
- cas pratique guidé ;
- mini-évaluation avant le test final.

```python
# ============================================================
# 4.10. Étape 4 bis — Recommandation de contenus par parcours et par niveau
# ============================================================

catalogue_contenus = {
    "DA": {
        "Débutant": [
            "Bases Excel",
            "Nettoyage simple des données",
            "Tableaux croisés dynamiques",
            "Graphiques simples"
        ],
        "Intermédiaire": [
            "Power Query",
            "Modèle de données Excel",
            "Jointures et relations",
            "Tableaux de bord simples"
        ],
        "Avancé": [
            "Power Pivot",
            "Mesures DAX de base",
            "Dashboard analytique",
            "Cas pratique complet DA"
        ]
    },
    "BI": {
        "Débutant": [
            "Introduction Power BI",
            "Importation des données",
            "Power Query débutant",
            "Visualisations simples"
        ],
        "Intermédiaire": [
            "Modèle en étoile",
            "Relations entre tables",
            "Mesures DAX",
            "Filtres et segments"
        ],
        "Avancé": [
            "DAX avancé",
            "Row-Level Security",
            "Optimisation du modèle",
            "Dashboard décisionnel"
        ]
    },
    "DS": {
        "Débutant": [
            "Bases Python",
            "Pandas pour CSV",
            "Nettoyage des données",
            "Statistiques descriptives"
        ],
        "Intermédiaire": [
            "Feature engineering",
            "Train/test split",
            "Classification et régression",
            "Évaluation des modèles"
        ],
        "Avancé": [
            "Random Forest",
            "XGBoost",
            "Optimisation des hyperparamètres",
            "Déploiement modèle avec FastAPI"
        ]
    },
    "IA": {
        "Débutant": [
            "Bases des LLM",
            "Prompt engineering",
            "System prompt",
            "Sorties structurées JSON"
        ],
        "Intermédiaire": [
            "RAG",
            "Embeddings",
            "Vector database",
            "Function calling"
        ],
        "Avancé": [
            "Agents IA",
            "MCP",
            "Workflow n8n",
            "Architecture IA complète"
        ]
    }
}

def recommander_contenus(row):
    parcours = row["parcours_recommande"]
    niveau = row["niveau_initial_global"]
    
    contenus = catalogue_contenus.get(parcours, {}).get(niveau, [])
    
    if niveau == "Débutant":
        priorite = "Haute"
    elif niveau == "Intermédiaire":
        priorite = "Moyenne"
    else:
        priorite = "Renforcement"
    
    return pd.Series({
        "contenus_recommandes": " ; ".join(contenus),
        "type_contenu": "Cours + exercices + mini-évaluation",
        "priorite_contenu": priorite,
        "raison_contenu": f"Contenus proposés selon le parcours {parcours} et le niveau {niveau}."
    })

df_recommandations_contenus = df_recommandations.copy()

resultats_contenus = df_recommandations_contenus.apply(
    recommander_contenus,
    axis=1
)

df_recommandations_contenus = pd.concat(
    [df_recommandations_contenus, resultats_contenus],
    axis=1
)

PATH_RECOMMANDATIONS_CONTENUS = OUTPUT_DIR / "07_recommandations_contenus_parcours_niveau.csv"

df_recommandations_contenus.to_csv(
    PATH_RECOMMANDATIONS_CONTENUS,
    index=False,
    encoding="utf-8-sig"
)

print("Recommandations de contenus générées :", df_recommandations_contenus.shape)
print("Fichier exporté :", PATH_RECOMMANDATIONS_CONTENUS)

display(df_recommandations_contenus[
    [
        "IDENTIFICATION",
        "niveau_initial_global",
        "parcours_recommande",
        "lacunes_detectees",
        "contenus_recommandes",
        "priorite_contenu",
        "raison_contenu"
    ]
].head(20))
```

# Cellule 12 — Ajustement après test intermédiaire

```python
# ============================================================
# 4.11. Étape 4 ter — Ajustement des contenus après test intermédiaire
# VERSION CORRIGÉE
# ============================================================

# Charger les fichiers nécessaires si les variables n'existent pas
INPUT_SCORE_DIR = Path("outputs/02_scoring_pedagogique")

PATH_SCORES_INTERMEDIAIRES = INPUT_SCORE_DIR / "02_scores_tests_intermediaires.csv"
PATH_CORRECTION_MI = INPUT_SCORE_DIR / "template_correction_tests_intermediaires.csv"

if "df_scores_intermediaires" not in globals():
    df_scores_intermediaires = pd.read_csv(PATH_SCORES_INTERMEDIAIRES)

if "df_correction_mi" not in globals():
    df_correction_mi = pd.read_csv(PATH_CORRECTION_MI)

print("Scores intermédiaires :", df_scores_intermediaires.shape)
print("Barème intermédiaire :", df_correction_mi.shape)


# Colonnes de score des questions intermédiaires
colonnes_score_mi = [
    col for col in df_scores_intermediaires.columns
    if col.startswith("score_mi_Q")
]

print("Colonnes scores MI détectées :", len(colonnes_score_mi))


# Mapping question standardisée -> question originale
mapping_questions_mi = (
    df_correction_mi
    .set_index(["parcours", "question_standardisee"])["question_originale"]
    .to_dict()
)


def detecter_lacunes_intermediaires(row):
    parcours = row["parcours_test_intermediaire"]
    lacunes = []
    
    for col_score in colonnes_score_mi:
        question = col_score.replace("score_", "")
        score = row[col_score]
        
        if score == 0:
            question_originale = mapping_questions_mi.get(
                (parcours, question),
                question
            )
            lacunes.append(question_originale)
    
    return lacunes


def recommander_remediation(row):
    lacunes = detecter_lacunes_intermediaires(row)
    niveau = row["niveau_intermediaire"]
    parcours = row["parcours_test_intermediaire"]
    
    if len(lacunes) == 0:
        return pd.Series({
            "lacunes_intermediaires_detectees": "Aucune lacune majeure",
            "contenu_remediation": "Renforcement avancé et cas pratique complet",
            "type_remediation": "Approfondissement",
            "priorite_remediation": "Faible",
            "raison_remediation": "Aucune erreur détectée au test intermédiaire."
        })
    
    if niveau == "Débutant":
        priorite = "Haute"
        type_remediation = "Remédiation obligatoire"
    elif niveau == "Intermédiaire":
        priorite = "Moyenne"
        type_remediation = "Remédiation ciblée"
    else:
        priorite = "Faible"
        type_remediation = "Renforcement ciblé"
    
    contenu = (
        f"Revoir les notions non maîtrisées du parcours {parcours}, "
        "faire des exercices ciblés, puis une mini-évaluation avant le test final."
    )
    
    return pd.Series({
        "lacunes_intermediaires_detectees": " | ".join(lacunes),
        "contenu_remediation": contenu,
        "type_remediation": type_remediation,
        "priorite_remediation": priorite,
        "raison_remediation": f"{len(lacunes)} lacune(s) détectée(s) au test intermédiaire."
    })


df_remediation_intermediaire = df_scores_intermediaires.copy()

resultats_remediation = df_remediation_intermediaire.apply(
    recommander_remediation,
    axis=1
)

df_remediation_intermediaire = pd.concat(
    [df_remediation_intermediaire, resultats_remediation],
    axis=1
)


PATH_REMEDIATION = OUTPUT_DIR / "08_remediation_apres_test_intermediaire.csv"

df_remediation_intermediaire.to_csv(
    PATH_REMEDIATION,
    index=False,
    encoding="utf-8-sig"
)

print("Remédiation intermédiaire générée :", df_remediation_intermediaire.shape)
print("Fichier exporté :", PATH_REMEDIATION)

print("\nRépartition des priorités de remédiation :")
display(df_remediation_intermediaire["priorite_remediation"].value_counts())

display(df_remediation_intermediaire[
    [
        "IDENTIFICATION",
        "parcours_test_intermediaire",
        "score_intermediaire_total",
        "niveau_intermediaire",
        "lacunes_intermediaires_detectees",
        "contenu_remediation",
        "priorite_remediation"
    ]
].head(20))
```
